## 1. Filter Function

Filter Function used to apply filter condition in nested json format columns

```sql
select 
     order_id,
     books,
     filter(books,x->x.subtotal> 34 ) as ExpensiveBooks
from order_vw

```

```sql
select 
      order_id,
      books,
      ExpensiveBooks
from (
    select 
     order_id,
     books,
     filter(books,x->x.subtotal> 34 ) as ExpensiveBooks
   from order_vw
)
where size(ExpensiveBooks)>0;
```

### Tax Calculator

```sql
CREATE OR REPLACE FUNCTION tax_calc(BaseSalary FLOAT)
RETURNS FLOAT
RETURN
    CASE
        WHEN BaseSalary > 5000 AND BaseSalary < 7000 THEN BaseSalary * 0.25
        WHEN BaseSalary >= 7000 AND BaseSalary < 10000 THEN BaseSalary * 0.35
        WHEN BaseSalary >= 10000 THEN BaseSalary * 0.45
        ELSE 0
    END;
```

```sql
select 
     id,
     EmpName,
     BaseSalary,
     tax_calc(BaseSalary) as tax
from employee
```

```sql
DESCRIBE function tax_calc
```

```sql
DESCRIBE FUNCTION extended tax_calc
```

```sql
select 
     id,
     empName,
     baseSalary,
     tax_calc(baseSalary) as tax,
     dense_rank() over (order by BaseSalary desc) as rank
from employee
```
